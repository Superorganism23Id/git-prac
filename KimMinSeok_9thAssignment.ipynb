### LAB 8-1 1번

```python
from datetime import datetime

now = datetime.now()

year = now.year
month = now.month
day = now.day
hour = now.hour
minute = now.minute
second = now.second

if hour < 12:
    am_pm = "오전"
    display_hour = hour
else:
    am_pm = "오후"
    display_hour = hour - 12

if display_hour == 0:
    display_hour = 12

print(f"오늘의 날짜 : {year}년 {month}월 {day}일")
print(f"현재시간 : {am_pm} {display_hour}시 {minute}분 {second}초")
```

설명 : datetime 모듈을 사용하여 오늘의 날짜와 현재 시간을 구하고, 년·월·일·시·분·초 형식으로 출력하는 문제.
현재 시간이 오전인지 오후인지 판단하여 오전/오후 형식으로 출력한다.

### LAB 8-2 1번

```python
from datetime import datetime

today = datetime.now()
christmas = datetime(2027, 12, 25)

left = christmas - today

days = left.days
hours = left.seconds // 3600

print(f"오늘은 {today.year}년 {today.month}월 {today.day}일입니다.")
print(f"2027년 크리스마스 까지는 {days}일 {hours}시간 남았습니다.")
```

설명 : datetime 모듈을 사용하여 오늘 날짜와 2027년 크리스마스 날짜의 차이를 계산하는 문제.
남은 기간을 일수와 시간 단위로 나누어 출력한다.

### LAB 8-2 3번

### LAB 8-3 1번

### LAB 8-4 3번

```python
import math

for degree in range(0, 181, 10):
    radian = math.radians(degree)
    value = math.sin(radian)
    print(f"sin({degree:3d}) = {value:.2f}")
```

설명 : datetime 모듈의 date.today()를 사용하여 오늘 날짜를 구하고, 연애 시작일인 2019년 2월 14일부터 오늘까지 며칠이 지났는지 계산하는 문제.
두 날짜를 뺄셈하여 경과한 날짜 수를 구한다

### 연습 8.1 2번

```python
from datetime import date, timedelta

start_day = date(2019, 2, 14)

day100 = start_day + timedelta(days=100)
day200 = start_day + timedelta(days=200)
day500 = start_day + timedelta(days=500)
day1000 = start_day + timedelta(days=1000)

print(f"춘향이와 몽룡이의 연애 시작일 : {start_day.year}년 {start_day.month}월 {start_day.day}일")
print(f"100일 기념일 : {day100.year}년 {day100.month}월 {day100.day}일")
print(f"200일 기념일 : {day200.year}년 {day200.month}월 {day200.day}일")
print(f"500일 기념일 : {day500.year}년 {day500.month}월 {day500.day}일")
print(f"1,000일 기념일 : {day1000.year}년 {day1000.month}월 {day1000.day}일")
```

설명 : 연애 시작일인 2019년 2월 14일을 기준으로 100일, 200일, 500일, 1000일 기념일을 계산하는 문제.
timedelta를 사용하여 시작 날짜에 원하는 날짜 수를 더한다.

### 연습 8.2 1번

```python
import time

def sum1to():
    total = 0

    for i in range(1, 1000001):
        total += i

    return total


start_time = time.time()

sum1to()

end_time = time.time()

print(f"1에서 1,000,000까지의 합을 구하는 시간 : {end_time - start_time:.4f}초")
```

설명 : 1부터 1,000,000까지의 정수 합을 구하는 함수 sum1to()을 작성하고, 이 함수가 한 번 실행되는 데 걸리는 시간을 측정하는 문제.
time 모듈의 time() 함수를 사용하여 시작 시간과 종료 시간의 차이를 구한다.

### 연습 8.2 2번

```python
import time

def sum1to():
    total = 0

    for i in range(1, 1000001):
        total += i

    return total


start_time = time.time()

for i in range(100):
    sum1to()

end_time = time.time()

print(f"1,000,000까지의 합을 100번 반복해서 구하는 시간 : {end_time - start_time:.4f}초")
```

설명 : sum1to() 함수를 100번 반복 실행하고, 전체 실행 시간을 측정하는 문제.
반복문을 사용하여 함수를 여러 번 호출한 뒤 총 소요 시간을 출력한다.

### 연습 8.3 1번

```python
import time

def hello_python():
    print("Hello Python!")


start_time = time.time()

for i in range(100):
    hello_python()

end_time = time.time()

print(f"Hello Python! 출력을 100번 반복하는 시간 : {end_time - start_time:.4f}초")
```

설명 : Hello Python!을 출력하는 함수를 작성하고, 이 함수를 100번 실행하는 데 걸리는 시간을 측정하는 문제.

### 연습 8.3 2번

```python
import time

def factorial10000():
    result = 1

    for i in range(1, 10001):
        result *= i

    return result


start_time = time.time()

for i in range(100):
    factorial10000()

end_time = time.time()

print(f"10000!을 100번 반복해서 구하는 시간 : {end_time - start_time:.4f}초")
```

설명 : 10000의 팩토리얼 값을 구하는 함수를 작성하고, 이 함수를 100번 실행하는 데 걸리는 시간을 측정하는 문제

### 연습 8.3 3번

```python
import time

def sum_odd_cube():
    total = 0

    for i in range(1, 10001, 2):
        total += i ** 3

    return total


start_time = time.time()

for i in range(100):
    sum_odd_cube()

end_time = time.time()

print(f"1에서 10000까지 홀수의 세제곱 더하기를 100번 반복하는 시간 : {end_time - start_time:.4f}초")
```

설명 : 1부터 10000까지의 모든 홀수를 세제곱한 뒤 더하는 함수를 작성하고, 이 함수를 100번 실행하는 시간을 측정하는 문제.

### 연습 8.3 4번

```python
import time
import math

def sum_sin():
    total = 0

    for degree in range(1, 361):
        radian = math.radians(degree)
        total += math.sin(radian)

    return total


start_time = time.time()

for i in range(100):
    sum_sin()

end_time = time.time()

print(f"1에서 360도까지 sin 값의 합을 100번 반복하는 시간 : {end_time - start_time:.4f}초")
```

설명 : 1도부터 360도까지의 각도에 대한 sin 값을 구하고, 그 합을 반환하는 함수를 작성하는 문제.
각도를 라디안으로 변환한 뒤 math.sin() 함수를 사용하며, 이 과정을 100번 반복하여 실행 시간을 측정.

### 연습 8.4 1번

```python
seed = 12234

def my_rand():
    global seed

    seed = (1103515245 * seed + 12345) % (2 ** 31)

    return seed % 1000001


for i in range(10):
    print(my_rand())
```

설명 : 0부터 1,000,000까지의 정수를 생성하는 의사난수 함수 my_rand()를 직접 작성하는 문제.
임의의 계수와 식은 교재를 참고하여 구현.

### 연습 8.4 2번

```python
import time

seed = int(time.time())

def my_rand():
    global seed

    seed = (1103515245 * seed + 12345) % (2 ** 31)

    return seed % 1000001


for i in range(2):
    for j in range(5):
        print(my_rand())
    print()
```

설명 : time 모듈의 time() 함수를 사용하여 현재 시간을 초기 시드값으로 설정하고, 의사난수를 생성하는 문제.
현재 시간은 실수값이므로 int()로 정수로 변환한 뒤 시드값으로 사용한다.
